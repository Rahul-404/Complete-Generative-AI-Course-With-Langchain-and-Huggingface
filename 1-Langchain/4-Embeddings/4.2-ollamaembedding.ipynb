## Ollama

Ollama supports embeddings models, making it possible to build retrival augmented generation (RAG) applications that combine text prompts with existing documents or other data.

```python
from langchain_community.embeddings import OllamaEmbeddings
```

```python
embeddings = (
    OllamaEmbeddings(model="mistral") ##by default it use ollama2
)
```

```python
embeddings
```

```python
r1 = embeddings.embed_documents(
    [
        "Alpha is the first letter of Greek alphabet",
        "Beta is the second letter of Greek alphabet"
    ]
)
```

```python
embeddings.embed_query("What is the second letter of Greek alphabet")
```
